1. **Лічильник до заданого числа.**
   Завдання:
     - Створіть цикл `while`, щоб він лічив від 1 до заданого числа в змінній `limit` (включно).
     - Виведіть кожне число, яке проходить лічильник, на екран.
     - В кінці виведіть повідомлення "Кінець розрахунку!".


```python
limit = 9  # Задане число
```

```python
# починаэмо лічильник з 1
count=1
# цикл працює, поки count не перевищить limit
while count<= limit:
  print(count)
  count+=1 # збільшуємо лічильник на 1
  print("Кінець розрахунку!")
```

2. **Пошук першого парного числа.** Завдання:
     - Створіть цикл `while`, щоб знайти перше парне число після заданого числа `start`.
     - Знайдіть та виведіть перше парне число.


```python
start = 7  # Задане число
```

```python
# починаємо перевірку з наступного числа після start
current=start+1
# шукаэмо перше парне число (число,яке ділиться на 2 без залишку)
while current % 2 !=0:
  current +=1
print(current)
```


3. **Обчислення факторіалу.**   
  Факторіал числа - це добуток всіх натуральних чисел від 1 до цього числа включно. Наприклад, факторіал числа 5 (позначається як 5!) дорівнює `1 * 2 * 3 * 4 * 5 = 120`.

  Завдання:  
      - Обчисліть факторіал заданого числа, використовуючи цикл `while`.  
      - Задайте деяке число для обчислення факторіалу у змінній `num`.  
      - Виведіть обчислений факторіал на екран.  


```python
# задаэмо число для обчислення факторыалу
num=5
# змінна для збереження результату (факторіал 0 та 1 дорівнює 1)
factorial=1
# лічильник починаємо з 1
i=1
#обчислюэмо факторыал за допомогою циклу while
while i<= num:
  factorial*=i # Множимо поточний результат на i
  i+=1 # збільшуємо лічильник
  print(factorial)
```
